```python
import numpy as np

class RBM:
  def __init__(self,n_visible,n_hidden):
    self.n_visible = n_visible
    self.n_hidden = n_hidden

    self.W=np.random.randn(n_visible,n_hidden)
    self.v_bias=np.zeros(n_visible)
    self.h_bias=np.zeros(n_hidden)


  def _sigmoid(self,x):
    return 1.0/(1.0+np.exp(-x))

  def propagate_up(self,v):
    return self._sigmoid(np.dot(v,self.W)+self.h_bias)

  def propagate_down(self,h):
    return np.dot(h,self.W.T)+self.v_bias

  def train(self,data,epochs,learning_rate):
    for epoch in range(epochs):
      h_prob_data=self.propagate_up(data)
      h_data=(h_prob_data > np.random.rand(h_prob_data.shape)).astype(float)

      v_reconstruction=self.propagate_down(h_data)
      v_reconstruction_binary=(v_reconstruction > np.random.rand(*v_reconstruction.shape)).astype(float)

      h_prob_reconstruction=self.propagate_up(v_reconstruction_binary)

      positive_grad=np.dot(data.T,h_prob_data)
      negative_grad=np.dot(v_reconstruction_binary.T, h_prob_reconstruction)

      self.W += learning_rate * (positive_grad - negative_grad)
      self.v_bias += learning_rate * np.sum(data-v_reconstruction_binary, axis=0)
      self.h_bias += learning_rate * np.sum(h_data-h_prob_reconstruction, axis=0)
```
